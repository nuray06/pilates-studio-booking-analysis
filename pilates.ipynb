```python
# Load CSV files into SQLite database as tables, creates connection, reads data, and imports into database
import sqlite3
import pandas as pd

conn = sqlite3.connect('C:/Users/saino/Downloads/pilates.db')

bookings = pd.read_csv('C:/Users/saino/Downloads/bookings.csv')
clients = pd.read_csv('C:/Users/saino/Downloads/clients.csv')

clients.to_sql('clients', conn, index=False, if_exists='replace')
bookings.to_sql('bookings', conn, index=False, if_exists='replace')

print("Data imported")
```

```python
# Join two tables into one DataFrame with SQL
query = """
SELECT 
    b.booking_id,
    b.class_type,
    b.instructor,
    b.date,
    b.time,
    b.status,
    c.client_id,
    c.first_name,
    c.last_name,
    c.gender,
    c.phone,
    c.email,
    c.join_date
FROM bookings b
JOIN clients c
ON b.client_id = c.client_id
"""

df = pd.read_sql_query(query, conn)
df.head()
```

```python
#Exporting the DataFrame into different file formats.
df.to_csv('Pilates.csv', index=False)
df.to_csv('Pilates.txt', index=False, sep='\t')
df.to_excel('Pilates.xlsx', index=False)
df.to_json('Pilates.json', orient='records', indent=4)
df.to_xml('Pilates.xml', index=False)
df.to_html('Pilates.html', index=False)
```

```python
# Data Structure Operations
# Demonstrate DataFrame concatenation along both axes (index and columns).
df1 = df[['client_id', 'booking_id']].head(5)
df2 = df[['class_type', 'status']].head(5)
# 1. Concatenate along index 
concat_row = pd.concat([df1, df1])
print(concat_row)
```

```python
# Concatenate along columns
concat_colm = pd.concat([df1, df2], axis=1)
concat_colm
```

```python
# Demonstrate all DataFrame column merging/joining types (Left join, index-on-index, and column-on-column).
# # Perform a left join on 'client_id' to combine information from two DataFrames
df1 = df[['client_id', 'first_name', 'last_name']].head(5)
df2 = df[['client_id', 'booking_id', 'class_type']].head(5)
left_joined = pd.merge(df1, df2, on='client_id', how='left')
left_joined
```

```python
#Join by index - index-on-index
index_joined = df1.join(df2, lsuffix='_left', rsuffix='_right')
index_joined
```

```python
#Merge by column - column-on-column
column_joined = pd.merge(df1, df2, left_on='client_id', right_on='client_id', how='inner')
column_joined
```

```python
# Melt
df_small = df[['client_id', 'booking_id', 'class_type', 'status']].head(5)
print(df_small)
df_melted = pd.melt(df_small, id_vars=['client_id'], value_vars=['booking_id', 'class_type', 'status'])
print("\nMelted DataFrame:")
print(df_melted)
```

```python
# pivot
df_pivot = df_melted.pivot(index='client_id', columns='variable', values='value')
print(df_pivot)
```

```python
# Group by class_type,status
grouped = df.groupby(['class_type', 'status'])
# Size of each group
print("Group sizes:\n", grouped.size())
# First booking_id in each group
print("\nFirst booking in each group:\n", grouped['booking_id'].first())
# Last booking_id in each group
print("\nLast booking in each group:\n", grouped['booking_id'].last())
```
